# RetailMind AI — Day 12: NLP & Classical Text Processing
**Objective**: Clean free-text product reviews, vectorize them using TF-IDF, and deploy a classical Logistic Regression baseline to extract customer sentiment.

```python
import joblib
model = joblib.load('../models/sentiment_baseline_v12.joblib')
test_reviews = ["Absolutely fantastic product, very durable.", "Horrible, fell apart on the first day."]
predictions = model.predict(test_reviews)
mapping = {1: 'Positive', 0: 'Negative'}
for review, pred in zip(test_reviews, predictions):
    print(f"Review: '{review}' \n--> Sentiment: {mapping[pred]}\n")
```
