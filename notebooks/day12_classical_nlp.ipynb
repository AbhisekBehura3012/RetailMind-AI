# RetailMind AI — Day 12: Classical NLP & Text Intelligence (TF-IDF + Naive Bayes)
**Objective**: Build classical bag-of-words / n-gram TF-IDF pipelines with Multinomial Naive Bayes for retail review sentiment polarity and automated support-ticket routing.

## 1. Load Persisted Day 12 TF-IDF Pipelines

```python
import joblib
import pandas as pd

sent_pipe = joblib.load('../models/tfidf_sentiment_v12.joblib')
ticket_pipe = joblib.load('../models/tfidf_ticket_v12.joblib')
print('Loaded Day 12 TF-IDF + MultinomialNB pipelines successfully.')
```

## 2. Inspect Top N-Gram Features & Run Sample Inference

```python
sample_texts = [
    'Amazing quality and fast shipping, works smoothly right out of the box!',
    'My package says delivered yesterday but nothing arrived, and I was charged twice.',
    'The zipper broke the very first time I wore the jacket.',
    'Cannot reset my password, verification email never arrives.'
]

sent_preds = sent_pipe.predict(sample_texts)
sent_probs = sent_pipe.predict_proba(sample_texts).max(axis=1)
ticket_preds = ticket_pipe.predict(sample_texts)
ticket_probs = ticket_pipe.predict_proba(sample_texts).max(axis=1)

df_results = pd.DataFrame({
    'Review / Ticket Text': sample_texts,
    'Sentiment': ['Positive' if p == 1 else 'Negative' for p in sent_preds],
    'Sentiment Conf': sent_probs.round(4),
    'Routed Queue': ticket_preds,
    'Routing Conf': ticket_probs.round(4)
})
df_results
```
